```python
import nltk
import spacy

from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
from vaderSentiment.vaderSentiment import SentimentIntensityAnalyzer

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('omw-1.4')

nlp = spacy.load("en_core_web_sm")

stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()
sentiment_analyzer = SentimentIntensityAnalyzer()
```

```python
text = "I love learning Natural Language Processing"

tokens = word_tokenize(text)

print("Tokenization:")
print(tokens)
```

```python
words = ["playing", "studies", "running", "happiness"]

print("Stemming:")

for word in words:
    print(word, ":", stemmer.stem(word))
```

```python
words = ["playing", "studies", "running", "better"]

print("Lemmatization:")

for word in words:
    print(word, ":", lemmatizer.lemmatize(word))
```

```python
text = "Rahul works at Google in Bangalore"

doc = nlp(text)

print("Named Entity Recognition:")

for entity in doc.ents:
    print(entity.text, ":", entity.label_)
```

```python
text = "I am very happy because I love Python programming"

score = sentiment_analyzer.polarity_scores(text)

print("Sentiment Scores:", score)

if score['compound'] >= 0.05:
    print("Sentiment: Positive")

elif score['compound'] <= -0.05:
    print("Sentiment: Negative")

else:
    print("Sentiment: Neutral")
```

```python
def nlp_chatbot():

    print("Welcome to NLP Chatbot!")

    while True:

        text = input("\nEnter your sentence: ")

        if text.lower() == "exit":
            print("Thank you!")
            break

        # Tokenization
        tokens = word_tokenize(text)
        print("\nTokenization:", tokens)

        # Stemming
        stemmed = [stemmer.stem(word) for word in tokens]
        print("\nStemming:", stemmed)

        # Lemmatization
        doc = nlp(text)
        print("\nLemmatization:")

        for token in doc:
            if not token.is_space:
                print(token.text, ":", token.lemma_)

        # NER
        print("\nNamed Entity Recognition:")

        for ent in doc.ents:
            print(ent.text, ":", ent.label_)

        # Sentiment Analysis
        score = sentiment_analyzer.polarity_scores(text)

        print("\nSentiment Analysis:")

        if score['compound'] >= 0.05:
            print("Positive")

        elif score['compound'] <= -0.05:
            print("Negative")

        else:
            print("Neutral")


nlp_chatbot()
```
